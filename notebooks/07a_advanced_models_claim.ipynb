## 07A Advanced Models for Claims

#### Purpose
- Develop the advanced model for claims

#### Steps: 
- LightGBM or XGBoost, tuning
- Claim extension: severity model and expected loss
- Calibration (isotonic or Platt)
- Select final models
- Save models and tuning logs
